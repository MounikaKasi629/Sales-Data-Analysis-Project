## 4.Sales Analysis

```python
total_sale=df["Sales_Amount"].sum()
print("Total Sale Amount=",total_sale)
avg_sale=df["Sales_Amount"].mean()
print("Average Sale Amount=",avg_sale)
max_sale=df["Sales_Amount"].max()
print("Maximum Sale=",max_sale)
min_sale=df["Sales_Amount"].min()
print("Minimum Sale=",min_sale)
```

### 4.1 Sales By Region

```python
sales_by_region=df.groupby("Region")["Sales_Amount"].sum()
print(sales_by_region.sort_values(ascending=False).head(1))
```

### 4.2 Sales By Product Category

```python
sales_by_product=df.groupby("Product_Category")["Sales_Amount"].sum()
print(sales_by_product.sort_values(ascending=False).head(1))
```

### 4.3 Sales By Sales Representative

```python
sales_by_rep=df.groupby("Sales_Rep")["Sales_Amount"].sum()
print(sales_by_rep.sort_values(ascending=False).head(1))
```

### 4.4 Payment Method Usage

```python
high_payment=df["Payment_Method"].value_counts()
print(high_payment.sort_values(ascending=False).head(1))
```

### 4.5 Sales Channel Usage

```python
high_sale_channel=df["Sales_Channel"].value_counts()
print(high_sale_channel.sort_values(ascending=False).head(1))
```

### 4.6 Sales By Customer Type

```python
total_sales_by_customer_type=df.groupby("Customer_Type")["Sales_Amount"].sum()
print(total_sales_by_customer_type.sort_values(ascending=False).head(1))
```

### 4.7 Average Sales By Customer Type

```python
avg_sales_by_cust_type=df.groupby("Customer_Type")["Sales_Amount"].mean()
print(avg_sales_by_cust_type.sort_values(ascending=False).head(1)) 
```

### 4.8 Monthly Sales Analysis

```python
df["Sales_Month"]=df["Sale_Date"].dt.month
total_sale_by_month=df.groupby("Sales_Month")["Sales_Amount"].sum()
print(total_sale_by_month.sort_values(ascending=False))
```

## Data Visualization
